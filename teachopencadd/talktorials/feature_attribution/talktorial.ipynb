## Explainable artificial intelligence (XAI) theory
### Definitions
**Explanations** Giving a clear and concise definition of explainable artificial intelligence (XAI) is difficult, because the term is used in many different contexts.
Instead, one can require that explanations satisfy certain properties (Jimenez-Luna et al., 2020). E.g., explanations should
- give a *human-interpretable* account of why a model produced a given output
- be *expressed in terms of meaningful entities* within medicinal chemistry or biology
- be *faithful* to the model's actual computation
- support *decision-making* or formulation of *testable hypotheses*
- ...

**XAI methods** Likewise, one can classify methods for obtaining explanations based on 
- *model-agnostic*, i.e., applicable to any model, or *model-specific*, i.e., tailored to a specific model architecture
- *post-hoc*, i.e., applied after the model has been trained, or *intrinsic*, i.e., built into the model architecture
- *local*, i.e., explaining individual predictions, or *global*, i.e., explaining the overall model behavior
- ...

Some of these concepts are explored in further detail using examples below.

### Basic concepts
#### Local vs. global explanations
Local explanations aim to explain individual predictions, for instance by highlighting which features were most influential for a specific instance. Global explanations, on the other hand, provide insights into the overall behavior of the model, identifying general patterns and important features across the entire dataset.

To limit the length of the practical section, this talktorial will focus on *local*, sometimes also called *instance-level* explanations. 

#### Faithfulness
**Definition and estimation** An explanation is faithful if it accurately reflects the model's internal mechanism that produced the prediction(s) (Jacovi & Goldberg, 2020). Faithfulness can be estimated by measuring how model predictions change when explanatory parts of the input are removed or modified. For instance, for molecules, if an explanation highlights certain atoms or fragments as important for a prediction, removing the corresponding atoms, or replacing them with chemically reasonable alternatives, should lead to a significant change in the model's output. This *deletion* (or *comprehensiveness*) protocol is the standard way faithfulness is quantified (Hooker et al., 2019; DeYoung et al., 2020).

Deleting parts of a molecule is not trivial, because the resulting structure must still be chemically valid. For instance, if we delete an atom, we must also delete all bonds connected to it. If this leaves a neighboring atom with an unsatisfied valence, we must fill these valencies, e.g. by inserting hydrogens. Even when such modifications are valid, they may still create chemically unrealistic molecules that are also out-of-distribution relative to the training data, making it harder to interpret corresponding model predictions.

**Implications**
Faithful explanations thus successfully recover features that are *causally* relevant to the trained model's prediction. If the training data contains spurious correlations, a faithful explanation will faithfully recover these correlations, which may not be meaningful to a human. This is why faithfulness is necessary but not sufficient for an explanation to be useful.


#### Plausibility
**Definition**
An explanation is plausible if it aligns with human intuition or domain knowledge (Jacovi & Goldberg, 2020; Jimenez-Luna et al., 2020). Plausibility and faithfulness are independent: an explanation can look chemically sensible while having nothing to do with the model's computation, and vice versa. *If* plausible (ground-truth) explanations are available, they can be used to quantitatively evaluate plausibility. Most of the time, however, plausibility can only be assessed qualitatively by human experts and only with a good amount of uncertainty, because the underlying mechanisms of molecular properties are often not fully understood.

**Plausible EGFR binding explanations**
In this talktorial we will study explanation plausibility on an example from the kinase domain. In mammals, kinases are a large family of enzymes that catalyze the transfer of phosphate groups from ATP to specific amino acid residues in proteins. Studies have shown that many kinases are also promising cancer drug targets (Sigismund et al., 2018) and led to the development of numerous FDA-approved kinase inhibitors. So called Type I kinase inhibitors bind competitively to ATP in the kinase's active site (van Linden et al., 2014; Kanev et al., 2021). In doing so most of them form at least one hydrogen bond with the hinge region of the kinase, which is a highly conserved structural motif that connects the N-terminal and C-terminal lobes of the kinase domain.

![test](images/hinge_binding_motif.png "Hinge")

The figure above highlights the hinge binding motif in ATP and the Type I kinase inhibitor Erlotinib. The binding motif in Erlotinib is a quinazoline ring with two nitrogen atoms that can form hydrogen bonds with the hinge region of the kinase and closely resembles the adenine ring of ATP. The hinge binding motif is present in many Type I kinase inhibitors.

When we generate explanations for model predictions that correctly assign a high pIC50 to a Type I kinase inhibitor, we can examine their plausibility qualitatively by checking whether the hinge binding motif (and especially the hydrogen-bond forming nitrogens) are among the features marked as important. It is worth being explicit about what that check can and cannot establish. If the hinge binding motif is *not highlighted* at all, it is safe to say the explanation is implausible. Being absolutely certain about plausibility is more difficult, because other substructures may also be vital for other non-covalent protein-ligand interactions that drive binding.


#### Applications of XAI in drug discovery
Three uses recur in the literature (Jimenez-Luna et al., 2020; Wellawatte et al., 2023), and they make different demands on the explanation.

**Model debugging and detecting spurious correlations** Attribution is often most useful when it looks *implausible*. If a model predicting EGFR affinity puts most of its weight on a solubilising side chain, or on a fragment that only ever appears in compounds from one publication, the likely cause is not chemistry but dataset structure: an assay artefact, a series of close analogues, a leak across the train/test split. This is the one application that does not need the explanation to be plausible, since it is the model's actual behaviour we want to see. To avoid wrongly blaming the model, however, the explanations should be as faithful as possible.

**Hypothesis generation** An attribution map is a cheap and specific proposal: *this* fragment is what the model believes drives the property. Because the claim is local and concrete, it can be tested, by re-predicting a small modification, or by making one. McCloskey et al. (2019) show both sides: attributions can recover a genuine binding mechanism, but only when the training data actually contains the causal signal, and they confidently report a wrong mechanism when it does not.

**Steering structure optimization** In a lead-optimization loop, per-atom attributions suggest *where* to modify a molecule: keep the positively attributed core, vary the negatively attributed periphery. Here the explanation guides a search rather than justifying a decision, so a method that is only approximately faithful is often good enough.  A wrong suggestion costs one wasted iteration, not a wrong conclusion.

#### Other concepts and further reading
For broader context beyond what this talktorial covers:

* **Surveys for drug discovery**: Jimenez-Luna et al. (2020) for XAI in drug discovery generally, and Wellawatte et al. (2023) for a perspective focused specifically on molecular property models.
* **Do we need post-hoc explanations at all?** Rudin (2019) argues that for high-stakes decisions one should prefer models that are interpretable by design over explaining black boxes after the fact.
* **Beyond attribution**: *counterfactual* explanations answer "what minimal change to the molecule would flip the prediction?" (Wellawatte et al., 2022), and *subgraph* explanations select a connected substructure rather than scoring atoms independently (Ying et al., 2019).


### Feature attribution methods
**Definition** Given a predictive model $f: \mathbb{R}^p \to \mathbb{R}$ and an input instance $x \in \mathbb{R}^p$ with $p$ features, a feature attribution method computes a vector of importance scores $\phi(x) = [\phi_1(x), \phi_2(x), ..., \phi_p(x)]$, where $\phi_i(x)$ represents the contribution of the $i$-th feature to the model's prediction $f(x)$ for the instance $x$. Thus these methods provide *local explanations* for the model's predictions. By aggregating attribution scores over multiple instances, one can also obtain more *global explanations* for the model's behavior.

**Visualization** In the case of molecules, it is common to analyze the feature attribution on the atom- and bond-level, which can be visualized by coloring the atoms and bonds in the molecule according to their importance using a gradient-based color map. This allows users to quickly identify which parts of the molecule may be driving the model's prediction.

#### Shapley values and TreeSHAP
**Definition** Shapley values (Shapley, 1953) are a concept from cooperative game theory that provide a fair way to distribute the total gain (or prediction) among the features based on their contributions. Their use as a feature-attribution method for machine learning models is due to Lundberg & Lee (2017), who showed that several earlier attribution methods are special cases of the same framework.
They are computed by considering all possible combinations of features and measuring the marginal contribution of each feature to the prediction. The Shapley value for a feature is the average of its marginal contributions across all possible feature subsets. The Shapley value for the $i$-th feature can be defined as

$$
\phi_i(x) = \sum_{S \subseteq N \setminus \{i\}} \frac{|S|! (|N| - |S| - 1)!}{|N|!} \left[ f_{S \cup \{i\}}(x_{S \cup \{i\}}) - f_S(x_S) \right].
$$

where $N$ is the set of all features, $S$ is a subset of features not including feature $i$, and $f_S(x_S)$ is the model's prediction when only the features in subset $S$ are present.

**Additivity** One nice property of Shapley values is *additivity* (sometimes also called completeness), which means that the sum of the Shapley values for all features equals the difference between the model's prediction and the expected value of the model's output over the training data:
$$
\sum_{i \in N} \phi_i(x) = f(x) - \mathbb{E}[f(X)]
$$

Note that additivity is necessary for faithful feature-attribution, but not sufficient. 

**TreeSHAP** Naively computing exact Shapley values is computationally expensive, especially for models with many features. However, for certain model types, such as tree-based models (e.g., random forests), efficient algorithms like TreeSHAP (Lundberg et al., 2018, 2020) have been developed to compute or approximate Shapley values in polynomial time.
TreeSHAP computes Shapley values by exploiting the structure of decision trees rather than enumerating all feature subsets. Instead of evaluating each coalition separately, it recursively walks each tree once while carrying a bookkeeping structure that tracks all subsets of the features along the current root-to-leaf path simultaneously, so the contribution of every subset is accumulated in a single pass. For an ensemble of $T$ decision trees with at most $L$ leaves and a maximum depth of $D$, this brings the cost down to polynomial time $\mathcal O(T\cdot L\cdot D^2)$, while still yielding exact Shapley values for the tree's value function, with attributions summing to the prediction minus the base value.

Note that "the tree's value function" is doing a lot of work in that sentence: $f_S(x_S)$ is not uniquely determined, and the choice matters. 
When using the [SHAP implementation](https://shap.readthedocs.io/en/latest/generated/shap.TreeExplainer.html), TreeSHAP's default `feature_perturbation="tree_path_dependent"` estimates a *conditional* expectation using the sample counts stored in the tree, whereas the `"interventional"` setting uses a background dataset to break dependence between features. The two give different attributions on correlated features -- and fingerprint bits are heavily correlated (Chen et al., 2020; Janzing et al., 2020).

#### Input x Gradient
**Definition** As you might be able to guess from the name, Input x Gradient (Shrikumar et al., 2016), a refinement of plain gradient saliency (Simonyan et al., 2014), requires the model being explained to be a function that is differentiable with respect to its input features. Apart from this, Input x Gradient is completely agnostic to the model architecture. The attribution score for the $i$-th feature is computed as the product of the input feature value and the gradient of the model's output with respect to that feature:

$$
\phi_i(x) = x_i \cdot \frac{\partial f(x)}{\partial x_i}
$$

Rather than just looking at the gradient, the method additionally rescales it by the magnitude of the input feature.

**Properties** When $f$ is a deep neural network, the gradient can be computed using backpropagation. This makes Input x Gradient very efficient to compute, as it only requires a single forward and backward pass through the model for each input instance. The biggest limitation of Input x Gradients is that faithfulness is not guaranteed and can be severely compromised in the presence of gradient saturation. A less efficient but more faithful alternative is Integrated Gradients (Sundararajan et al., 2017), which computes the average gradient along a straight path from a baseline input $\bar x$ to the actual input, thus avoiding the problem of gradient saturation. Integrated Gradients also satisfies completeness with respect to that baseline -- but it forces you to pick one, and there is no obvious "empty molecule" to use. Hence it is not further explored in this talktorial.

## Model training


### Load and split raw data
First, we load the raw data using [polars](https://docs.pola.rs/) and add a column that randomly splits data into train, validation and test sets. The split is reproducible by setting a random seed.

We also rewrite every SMILES string into its RDKit canonical form. That looks like cosmetics, but it is what makes the explanations comparable later on: an attribution is just an array indexed by atom position, and several parts of this notebook re-parse the SMILES independently (the fingerprint generator, the chemprop featurizer, the drawing code). Canonicalizing once, up front, makes "atom 7" refer to the same atom everywhere. Only 4 of the 6 457 input SMILES are not already canonical -- which is precisely why the resulting bug would have been hard to find.

We also remove named quinazoline-based kinase inhibitos (erlotinib, gefitinib, lapatinib) used later on for plausibility checks from the model training data.

```python
# define an expression that adds a random split column to any data frame
rand_unif = pr.rand(0, 1, seed=SEED)
random_split = (
    pl.when(rand_unif < FRAC_TEST)
    .then(pl.lit("test"))
    .when(rand_unif.is_between(FRAC_TEST, FRAC_TEST + FRAC_VAL))
    .then(pl.lit("val"))
    .otherwise(pl.lit("train"))
)

# canonical SMILES, so that every re-parse of the string gives the same atom order
canonical_smiles = pl.col(SMILES).map_elements(
    lambda smiles: Chem.MolToSmiles(Chem.MolFromSmiles(smiles)), return_dtype=pl.String
)

# load, split and canonicalize the data
df_egfr = (
    pl.scan_csv(DATA / "raw" / "egfr.csv")
    .with_columns(random_split.alias(SPLIT))
    .collect()
    .with_columns(canonical_smiles.alias(SMILES))
)

# force quinazoline inhibitors into the test data
df_egfr = df_egfr.with_columns(
    pl.when(pl.col.chembl_compound_id.is_in(EXCLUDE_CHEMBL_COMPOUNDS))
    .then(pl.lit("test"))
    .otherwise(pl.col.random_split)
    .alias(SPLIT)
)

# inspect absolute data split sizes
df_egfr.group_by(SPLIT, maintain_order=True).agg(pl.len())
```

Below we also verify that the quinazoline motif does show up in our training data, so that the model has a chance to learn to use it if it is a discriminative feature.

```python
_quinazoline = Chem.MolFromSmiles(QUINAZOLINE)


def match_quinazoline(smiles: str) -> bool:
    """Check if a SMILES string contains a quinazoline substructure."""
    mol = Chem.MolFromSmiles(smiles)
    matches = mol.GetSubstructMatches(_quinazoline)
    return bool(matches)


contains_quinazoline = (
    pl.col(SMILES)
    .map_elements(match_quinazoline, return_dtype=pl.Boolean)
    .alias("contains_quinazoline")
)

y = pl.col.pIC50
df_egfr.with_columns(contains_quinazoline).group_by(
    "contains_quinazoline", maintain_order=True
).agg(
    pl.len(),
    y.quantile(0.25).alias("q25-pIC50"),
    y.quantile(0.5).alias("median-pIC50"),
    y.quantile(0.75).alias("q75-pIC50"),
)
```

### Chemprop: dataset preparation
Next, we prepare the data by creating chemprop `MoleculeDataset` objects for the train, validation and test sets using the default chemprop featurizer.
Targets (pIC50 labels) are normalized to zero mean and unit variance using the training set statistics to stabilize training of the neural network.

```python
# polars data frame to chemprop dataset
def to_chemprop_dataset(data_frame: pl.DataFrame):
    datapoints = [
        MoleculeDatapoint.from_smi(smiles, np.array([y]))
        for smiles, y in data_frame.select(SMILES, TARGET).iter_rows()
    ]
    return MoleculeDataset(datapoints, featurizer=FEATURIZER)
```

```python
train_dataset = to_chemprop_dataset(df_egfr.filter(pl.col(SPLIT) == "train"))
scaler = train_dataset.normalize_targets()

val_dataset = to_chemprop_dataset(df_egfr.filter(pl.col(SPLIT) == "val"))
val_dataset.normalize_targets(scaler)

test_dataset = to_chemprop_dataset(df_egfr.filter(pl.col(SPLIT) == "test"))

output_transform = UnscaleTransform.from_standard_scaler(scaler)

train_loader = build_dataloader(train_dataset, batch_size=BATCH_SIZE, num_workers=0)
val_loader = build_dataloader(val_dataset, batch_size=BATCH_SIZE, num_workers=0, shuffle=False)
test_loader = build_dataloader(test_dataset, batch_size=BATCH_SIZE, num_workers=0, shuffle=False)
```

```python
# helper function for making a prediction on a single molecule given as smiles
def single_prediction(
    smiles: str,
    model: MPNN,
):
    dataset = MoleculeDataset([MoleculeDatapoint.from_smi(smiles)], featurizer=FEATURIZER)
    loader = build_dataloader(dataset, batch_size=1)
    with torch.inference_mode():
        model.eval()
        for batch in loader:
            result = model.predict_step(batch, 0)
            return result.item()
```

### Chemprop: model training and testing
Chemprop implements it's models using the [lightning](https://lightning.ai/) framework. We thus use the lightning `Trainer` class to run model training and validation. The trainer handles the training loop, including forward and backward passes, optimization, and evaluation on the validation set. We set all chemprop hyperparameters to their default values, as hyperparameter optimization is out of scope for this talktorial. We only set the number of epochs to a small, but sufficient value to ensure reasonable convergence of the model, while avoiding unfeasible training times.

```python
SKIP_TRAINING = False
ALLOW_OVERWRITE = True
```

```python
trainer = torch_lightning.Trainer(
    logger=True,
    enable_progress_bar=True,
    enable_checkpointing=False,
    accelerator="cpu",
    max_epochs=15,  # number of epochs to train for
)

# only train readout, use frozen chemeleon message-passing embedding
message_passing = load_pretrained_chemeleon()
message_passing.requires_grad_(False)
chemprop_model = MPNN(
    message_passing,
    MeanAggregation(),
    RegressionFFN(
        input_dim=message_passing.output_dim,
        output_transform=output_transform,
    ),
    batch_norm=True,
    metrics=[RMSE(), R2Score()],
)
```

### Random forest: dataset preparation
The random forest model will be trained on Morgan (ECFP) fingerprints, which are computed using RDKit. These are binary vector representations of molecules that capture the presence or absence of specific substructures. To ensure a fixed-length representation, most implementations rely on a hashing function that groups substructures into a fixed number of bits. This kind of fingerprint is not invertable, but one can store which atoms contributed to which bits, allowing for a mapping of fingerprint feature attributions back to the molecular graph.

To leverage this principle, we implement a helper class that generates morgan fingerprints and the inversion mapping from fingerprint bits to atoms for given SMILES strings in [src/talktorial_xai/util/fingerprints.py](src/talktorial_xai/util/fingerprints.py). This class is used to prepare the training, validation and test datasets for the random forest model below.

```python
fpGen = InvertibleFingerprintGen()


X_train, inv_maps_train = fpGen.batch(df_egfr.filter(pl.col(SPLIT) == "train")[SMILES])
X_test, inv_maps_test = fpGen.batch(df_egfr.filter(pl.col(SPLIT) == "test")[SMILES])

y_train = df_egfr.filter(pl.col(SPLIT) == "train")[TARGET].to_numpy()
y_test = df_egfr.filter(pl.col(SPLIT) == "test")[TARGET].to_numpy()

# inspect data shapes
print(f"Train data shape: {X_train.shape}")
print(f"Number of inversion maps: {len(inv_maps_train)}")
print(f"Shape of first inversion map: {inv_maps_train[0].shape}")
# memory intensive object
del inv_maps_train
```

Note that the inversion mappings are stored as a list of numpy matrices, one for each molecule of shape `(num_bits, num_atoms)`, where `num_atoms` is the number of atoms in the molecule and `num_bits` is the length of the fingerprint vector. Each entry in the matrix indicates whether a particular atom contributed to a specific fingerprint bit. 

Using a randomly chosen molecule from the EGFR dataset, we visualize the first three non-zero rows of its inversion map below. This shows precisely which atoms are responsible for setting these bits to the on state.

We will use the same mapping to decide which atoms to highlight when visualizing Morgan feature attribution on a molecular graph.

```python
sample_smiles = df_egfr.sample(1, seed=SEED).select("smiles").item()
_, inv_map = fpGen.get(sample_smiles)
# first 3 on bits only
panel = dict()
for idx, contributing_atoms_row in enumerate(inv_map):
    if not contributing_atoms_row.any():
        continue
    panel[f"Atoms contributing to bit #{idx}"] = Attribution(
        sample_smiles, None, contributing_atoms_row.astype(float)
    )
    if len(panel) == 4:
        break

_ = draw_attribution_panel(panel, alpha=0.6, colorbar_label="")
```

### Random forest: model training and testing
We train a random forest model using the scikit-learn implementation. The model is trained on the Morgan fingerprints generated in the previous step. Again, since this is out of scope, we do not do any further data preprocessing or hyperparameter optimization, and simply use the default hyperparameters of the scikit-learn implementation. We only limit the maximum tree depth, as TreeSHAP runtime scales quadratically with the depth of the trees, and we want to keep the runtime of this talktorial reasonable.

Limiting tree depth can make the model underfit, but we found that a maximum depth of 20 is sufficient to achieve reasonable performance on the EGFR dataset.

```python
rf_model = RandomForestRegressor(max_depth=20, random_state=SEED)
_ = rf_model.fit(X_train, y_train)
```

Again, we use the $R^2$ score and RMSE to evaluate the model's performance on the test data.

```python
pred_test = rf_model.predict(X_test)
r2 = r2_score(y_test, pred_test)
rmse = root_mean_squared_error(y_test, pred_test)
print(f"R^2  = {r2:.3f}")
print(f"RMSE = {rmse:.3f}")
```

Both models land in the same place: $R^2 \approx 0.68$–$0.70$ and an RMSE of about $0.7$ pIC50 units on the held-out test set, i.e. a typical prediction is off by roughly a factor of five in IC50. Chemprop is marginally ahead, but the gap is much smaller than the spread one would see from re-splitting the data, so it should not be read as one architecture beating the other -- particularly since we froze the CheMeleon encoder and trained only the readout, capped the forest's depth at 20, and tuned no hyperparameters for either model.

For our purposes this near-tie is the useful outcome. Two models of comparable accuracy built on entirely different representations (hashed circular substructures on one side, learned message passing over the molecular graph on the other) give us a controlled setting: any disagreement between their *explanations* cannot simply be blamed on one of them being the worse predictor.

## Explaining model predictions

### Visualization code
We implement a dataclass for storing any kind of attribution-based explanation for a molecular property prediction, irrespective of whether it was generated using TreeSHAP or Input x Gradient in [attribution.py](src/talktorial_xai/attribution.py).

Two functions in [visualization.py](src/talktorial_xai/visualization.py) do the drawing:

- **`draw_atom_attributions(mol, atom_weights, ...)`** draws a *single* molecule. It colours atoms -- and, when the method provides them, bonds -- with a diverging colormap that is deliberately *symmetric around zero*, so that red and blue always mean "pushed the prediction up" and "pushed it down", never merely "above or below this molecule's own average". It returns a matplotlib figure with a colour bar, and it raises if the number of weights does not match the number of atoms, which is the cheapest available guard against silently plotting one molecule's attributions onto another.
- **`draw_attribution_panel({title: attribution}, ...)`** draws *several* of those side by side. An `Attribution` carries the molecule it was computed on, so the structure is not passed again; a `(mol, attribution)` pair is still accepted for the case where it has `smiles=None` or should be drawn on a different depiction. Its one argument worth knowing is `shared_scale`: left at its default `True`, every panel uses one common colour scale and one colour bar, which is the only setting under which comparing panels by eye is valid. Set it to `False` when the magnitudes are incomparable anyway -- for instance when one panel holds TreeSHAP values in pIC50 units and the next holds Input x Gradient values in arbitrary units.

Both accept a `bonds` policy (`"own"`, `"fold"` or `"ignore"`) that decides what to do with bond attributions, since only some methods produce them; `Attribution.plot` documents the three options.

The cell below feeds the plotting code an attribution whose answer we already know, so that anything surprising in the picture is the plot's fault and not the model's.

```python
# A "model" whose explanation we know in advance: every non-carbon atom scores
# +1, every carbon 0. Nothing is learned here -- the point is to see what the
# plotting code does with an answer we can verify by looking at the structure.
dummy_smiles = df_egfr.sample(1, seed=SEED).select(SMILES).item()
dummy_mol = Chem.MolFromSmiles(dummy_smiles)
heteroatom_weights = np.array([float(atom.GetSymbol() != "C") for atom in dummy_mol.GetAtoms()])

dummy_attribution = Attribution(
    smiles=dummy_smiles,
    prediction=float(heteroatom_weights.sum()),
    atom_attributions=heteroatom_weights,
)
_ = dummy_attribution.plot(
    title="dummy attribution: heteroatoms only",
    subtitle="every non-carbon atom scores +1, every carbon 0",
    prediction_label="number of heteroatoms",
    annotate_top_k=3,
    alpha=0.7,
)
```

### Random forest: TreeSHAP
As mentioned above, we use TreeSHAP to compute feature attributions for the random forest model.

We implement a function [tree_shap](src/talktorial_xai/tree_shap.py) that computes TreeSHAP attributions for a given random forest model and a set of input instances.
`tree_shap(molecules, model, ...)` does three things:

1. It obtains the fingerprint matrix `X` together with the bit-to-atom incidence maps -- either by fingerprinting SMILES itself, or, as we do below, by accepting the `(X, inv_maps)` pair that was already built for training, so the bits are guaranteed to mean what they meant at fit time.
2. It runs `shap.TreeExplainer` on `X`, giving one SHAP value per *fingerprint bit*.
3. It pushes those per-bit values back onto atoms through the incidence maps and returns one `Attribution` per molecule, in input order.

Step 3 is where the only real decision lies:
each set bit stands for one or more circular atom environments, so we can push its
SHAP value back onto the atoms it covers, splitting the value equally over them. This does **not** preserve completeness at the level of atoms. A bit that is *off*
still carries a SHAP value: the absence of a substructure is evidence too, but it
has no environment, hence no atoms to receive it. That mass is a per-molecule constant, reported separately as `unattributed` instead of silently dropped.

```python
# passing the SMILES along is what lets us assert -- rather than assume -- that
# the two models' attributions refer to the same molecules and the same atoms
test_smiles = df_egfr.filter(pl.col(SPLIT) == "test")[SMILES]
rf_attributions = tree_shap((X_test, inv_maps_test), rf_model, smiles=test_smiles)
```

### Chemprop: Input x Gradient
For the D-MPNN the input is not a flat vector, so "the gradient with respect to the input" needs a choice of which input. We implement a function [input_x_gradient](src/talktorial_xai/input_x_gradient.py), which takes it with respect to the two featurized tensors that enter message passing:

- `bmg.V`, atom features, shape `(atoms in batch, atom_feature_dim)`
- `bmg.E`, bond features, shape `(2 x bonds in batch, bond_feature_dim)`, one row per directed edge

`input_x_gradient(molecules, model, ...)` then does three things:

1. It featurizes the molecules, or, as below, accepts the MoleculeDataset we already built, so the graphs the gradients flow through are exactly the ones the model was evaluated on.
2. It runs one forward pass and a single backward pass per batch. Each molecule's score depends only on its own nodes and edges, so backpropagating the summed batch score recovers every molecule's gradients at once.
3. It multiplies element-wise by the inputs, sums over the feature axis, and returns one Attribution per molecule with both an atom and a bond channel; the two directed edges of a bond are summed onto that bond.

```python
chemprop_attributions = input_x_gradient(test_dataset, chemprop_model)
```

### Comparing explanations across models
Both models now have one attribution per atom for every test molecule, and the two lists are in the same order and we can look into visualing and comparing their explanations.

#### **Visually**
Below, we sample a random molecule and plot the prediction and the two attributions side by side. 
Models often agree on the prediction, but disagree on the explanation.

The chemprop Input x Gradient attribution is more diffuse, with many atoms contributing a small amount to the prediction. The random forest TreeSHAP attribution is more concentrated, where oftentimes fewer of the present atoms contribute to the model prediction.

Considering the scales of attribution on the colorbar ticks is also important: the TreeSHAP explanations put a lot of mass on the absence of substructures, which is natural given the sparsity of Morgan fingerprints.
Hence, the atom attributions which collect only SHAP values from bits that are present are comparatively small.

```python
i = np.random.randint(0, len(chemprop_attributions), size=1)[0]

_ = draw_attribution_panel(
    {
        "Chemprop": chemprop_attributions[i],
        "Random Forest": rf_attributions[i],
    },
    bonds="fold",
    shared_scale=False,
)
```

#### **Quantifying aggreement**

```python
from scipy.stats import spearmanr

rank_correlations = []
for rf_attribution, chemprop_attribution in zip(
    rf_attributions, chemprop_attributions, strict=True
):
    # same molecule, same atom order -- checked, not assumed
    if rf_attribution.smiles != chemprop_attribution.smiles:
        raise ValueError(
            f"molecule mismatch: {rf_attribution.smiles} vs {chemprop_attribution.smiles}"
        )
    rf_atoms = rf_attribution.atom_attributions
    # TreeSHAP has no bond channel, so fold chemprop's bond values into its atoms
    chemprop_atoms = chemprop_attribution.folded_atom_attributions()
    if rf_atoms.shape != chemprop_atoms.shape:
        raise ValueError(f"atom count mismatch for {rf_attribution.smiles}")

    rho = spearmanr(rf_atoms, chemprop_atoms).statistic
    if np.isfinite(rho):  # undefined if either side is constant, e.g. for benzene
        rank_correlations.append(rho)

rank_correlations = np.array(rank_correlations)
print(f"molecules compared:    {len(rank_correlations)} / {len(rf_attributions)}")
print(f"mean Spearman rho:     {rank_correlations.mean():+.3f}")
print(f"median Spearman rho:   {np.median(rank_correlations):+.3f}")
print(f"fraction with rho > 0: {(rank_correlations > 0).mean():.1%}")
```

##### **Quinazoline-based type I inhibitors**

To study plausibility of the explanations, we will look at the model predictions and explanations for a set of quinazoline-based type I kinase inhibitors. These molecules are known to bind to the hinge region of the kinase domain, and thus we expect that a plausible explanation should highlight the hinge-binding motif as important for the model's prediction.

We will look at the results for the inhibitors themselves and modified versions where the nitrogen atoms in the hinge-binding quinazoline motif are replaced with carbon atoms.

Below, we define a simple helper function and the case study molecules.

```python
def strip_quinazoline_nitrogens(smiles: str) -> str:
    """Strips nitrogens from the quinazoline scaffold in a SMILES string.

    Args:
        smiles (str): SMILES string of the molecule.

    Raises:
        ValueError: If the SMILES string is invalid.
        ValueError: If no quinazoline scaffold is found in the molecule.

    Returns:
        str: SMILES string of the molecule with quinazoline nitrogens stripped.
    """
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        raise ValueError(f"Invalid SMILES: {smiles!r}")

    matches = mol.GetSubstructMatches(Chem.MolFromSmiles(QUINAZOLINE))
    if not matches:
        raise ValueError(f"No quinazoline scaffold in {smiles!r}")

    rw = Chem.RWMol(mol)
    for idx in {i for m in matches for i in m}:
        a = rw.GetAtomWithIdx(idx)
        if a.GetAtomicNum() == 7:
            a.SetAtomicNum(6)
            a.SetFormalCharge(0)
            a.SetNoImplicit(False)
            a.SetNumExplicitHs(0)

    out = rw.GetMol()
    Chem.SanitizeMol(out)
    return Chem.MolToSmiles(out)


compound_names = ["Gefitinib", "Erlotinib", "Lapatinib"]
compound_smiles = [GEFITINIB, ERLOTINIB, LAPATINIB]


def decorate_intervention(names):
    return [f"{name} (no-nitrogen)" for name in names]
```

#### **Unmodified inhibitors**

```python
chemprop_attr = input_x_gradient(
    compound_smiles,
    chemprop_model,
    featurizer=FEATURIZER,
    batch_size=1,
)
rf_attr = tree_shap(compound_smiles, rf_model, fp_gen=fpGen)
```

```python
_ = draw_attribution_panel(
    dict(zip(compound_names, chemprop_attr)),
    shared_scale=True,
    bonds="ignore",
    suptitle="Chemprop Input x Gradient",
)
```

**Random forest TreeSHAP**

```python
_ = draw_attribution_panel(
    dict(zip(compound_names, rf_attr)),
    shared_scale=True,
    bonds="ignore",
    suptitle="Random Forest TreeSHAP",
)
```
